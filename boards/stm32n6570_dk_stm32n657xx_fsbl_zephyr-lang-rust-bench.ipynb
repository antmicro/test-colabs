## Start Renode

```python
import os
from renode_run import get_default_renode_path
from renode_run.utils import RenodeVariant

os.environ['PYRENODE_RUNTIME'] = 'coreclr'
os.environ['PYRENODE_PATH'] = str(get_default_renode_path(variant=RenodeVariant.DOTNET_PORTABLE))

from pyrenode3.wrappers import Emulation, Monitor, TerminalTester, LEDTester
from Antmicro.Renode.Peripherals.UART import UARTBackend
from Antmicro.Renode.Analyzers import LoggingUartAnalyzer
from System import String

currentDirectory = os.getcwd()
emulation = Emulation()
monitor = Monitor()
emulation.BackendManager.SetPreferredAnalyzer(UARTBackend, LoggingUartAnalyzer)
```

## Setup a script

```python
%%writefile script.resc
logFile $ORIGIN/zephyr-lang-rust-bench-renode.log True

$name?="stm32n6570_dk_stm32n657xx_fsbl"
$bin?=@https://zephyr-dashboard.renode.io/zephyr/bc5d59d1b6657cfeef30124b818c8fbd69da68ac/stm32n6570_dk_stm32n657xx_fsbl/zephyr-lang-rust-bench/zephyr-lang-rust-bench.elf
$repl?=$ORIGIN/zephyr-lang-rust-bench.repl

using sysbus
mach create $name

machine LoadPlatformDescription @https://zephyr-dashboard.renode.io/zephyr_sim/bc5d59d1b6657cfeef30124b818c8fbd69da68ac/03e7bf4e259b85bdc1bf5dc63c4053dcc7c54680/stm32n6570_dk_stm32n657xx_fsbl/zephyr-lang-rust-bench/zephyr-lang-rust-bench.repl
machine EnableProfiler $ORIGIN/metrics.dump



showAnalyzer usart1

usart1 RecordToAsciinema $ORIGIN/zephyr-lang-rust-bench-asciinema
set osPanicHook
"""
self.ErrorLog("OS Panicked")
"""
cpu0 AddSymbolHook "z_fatal_error" $osPanicHook


macro reset
"""
    sysbus LoadELF $bin 
    cpu0 VectorTableOffset `sysbus GetSymbolAddress "_vector_table"`
    cpu0 EnableZephyrMode
    cpu0 EnableProfilerCollapsedStack $ORIGIN/zephyr-lang-rust-bench-profile true 62914560 maximumNestedContexts=10
"""

runMacro $reset
```

## Run the sample

```python
monitor.execute_script(currentDirectory + "/script.resc")
machine = emulation.get_mach("stm32n6570_dk_stm32n657xx_fsbl")
terminalTester = TerminalTester(machine.sysbus.usart1, 60)

terminalTester.WaitFor(String("Done with all tests"), pauseEmulation=True)

emulation.Dispose()
```
